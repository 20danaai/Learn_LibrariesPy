```python
import torch
```

```python
random_tensor=torch.rand(size=(2,3))
random_tensor,random_tensor.shape
```

```python
r_t_transpose=torch.transpose(random_tensor,dim0=0,dim1=1)
r_t_transpose,r_t_transpose.shape
```

```python
r_t_v2=random_tensor.t()
r_t_v2,r_t_v2.shape
```

```python
tensor_mean=torch.arange(0,16,2)
tensor_mean,tensor_mean.dtype
```

```python
tensor_mean_float=tensor_mean.type(torch.float32) # mean is only used with float
tensor_mean_float,tensor_mean_float.dtype,tensor_mean_float.shape
```

```python
float_mean=tensor_mean_float.mean()
float_mean
```

```python
reshaped=tensor_mean_float.reshape(2,4,1,-1) # to change the dim , the product must equal the original num of ele
reshaped,reshaped.shape,reshaped.ndim
```

```python
reshaped_v2=tensor_mean_float.reshape(1,-1)
reshaped_v2,reshaped_v2.shape,reshaped_v2.ndim
```

```python
x=torch.rand(size=(3,1))
x,x.shape,x.ndim
```

```python
x_stack=torch.stack([x,x,x])
x_stack,x_stack.shape,x_stack.ndim
```

```python
x_squeeze=x.squeeze() # delet dim
x_squeeze,x_squeeze.shape,x_squeeze.ndim
```

```python
x_unsqueeze=x.unsqueeze(dim=0) # add dim in first
x_unsqueeze,x_unsqueeze.shape,x_unsqueeze.ndim
```

```python
x_unsqueeze_v2=x.unsqueeze(dim=1) # add dim in end
x_unsqueeze_v2,x_unsqueeze_v2.shape,x_unsqueeze_v2.ndim
```

```python
i=torch.rand(224,228,3)
```

```python
i_permute=torch.permute(i,(2,0,1)) # change the order
i_permute.shape,i_permute.ndim
```
